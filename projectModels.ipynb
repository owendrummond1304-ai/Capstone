```python
# Train-Test split and sclaing x variables

x_train_g, x_test_g, y_train_g, y_test_g = train_test_split(x_g, y_g, test_size=0.2, random_state=42)

scaler = StandardScaler()

x_train_g_scaled = scaler.fit_transform(x_train_g)
x_test_g_scaled = scaler.transform(x_test_g)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.ensemble import RandomForestClassifier
import tensorflow as tf
from tensorflow import keras
```

```python
# Initialization of models and setup for cross-fold validation

# XGB

xgb_model = XGBClassifier(
    random_state = 42,
    eval_metric = 'auc',
    objective='binary:logistic',
)

xgb_param_dist = {
    'n_estimators': [500, 1000],
    'learning_rate': [0.01, 0.05],
    'max_depth': [3, 5, 7, 9, 11]
}

xgb_random = RandomizedSearchCV(
    xgb_model,
    param_distributions=xgb_param_dist,
    n_iter=10,
    scoring='roc_auc',
    cv=5,
    verbose=1,
    n_jobs=-1,
    random_state=42
)


## Light GBM

gbm_model = LGBMClassifier(
    random_state = 42
)

gbm_param_dist = {
    'n_estimators': [500, 1000],
    'learning_rate': [0.01, 0.05],
    'num_leaves': [31, 50, 100],
    'max_depth': [-1, 10, 20]
}

gbm_random = RandomizedSearchCV(
    gbm_model,
    param_distributions=gbm_param_dist,
    n_iter=10,
    scoring='roc_auc',
    cv=5,
    verbose=1,
    n_jobs=-1,
    random_state=42
)


# Random Forest Classifier


rf_model = RandomForestClassifier(
    random_state=42
)

rf_param_dist = {
    'n_estimators': [200, 400, 600],
    'max_depth': [None, 10, 20, 30, 40],
    'min_samples_split': [2, 5, 10],
    'max_features': ['sqrt', 'log2']
}

rf_random = RandomizedSearchCV(
    rf_model,
    param_distributions=rf_param_dist,
    n_iter=10,
    scoring='roc_auc',
    cv=5,
    verbose=1,
    n_jobs=-1,
    random_state=42
)
```

```python
# Cross-validation training models of all types and retaining accuracy of all models trained.

models = {
    "RandomForest": rf_random,
    "XGBoost": xgb_random,
    "LightGBM": gbm_random
}

results = {}

for name, model in models.items():
    print(f"Training {name}...")
    
    if name == 'RandomForest' or name == 'XGBoost' or name == 'LightGBM':
        model.fit(x_train_g_scaled, y_train_g)
        results[name] = model.best_score_
        print(f"{name} Best ROC-AUC: {model.best_score_:.4f}")
```

```python
from sklearn.metrics import roc_auc_score

test_probabilities = {}

print("--- FINAL EVALUATION ---")

# Get individual Test Scores
for name, search_object in models.items():
    # Retrieve the best version found during the search
    best_model = search_object.best_estimator_
    
    # Predict the probability of a Home Win
    probs = best_model.predict_proba(x_test_g_scaled)[:, 1]
    
    test_probabilities[name] = probs

    auc = roc_auc_score(y_test_g, probs)
    print(f"{name} Final Test AUC: {auc:.4f}")

# Get the Ensemble Score
prob_df = pd.DataFrame(test_probabilities)
ensemble_probs = prob_df.mean(axis=1)

ensemble_auc = roc_auc_score(y_test_g, ensemble_probs)

print(f"TRIPLE ENSEMBLE Final Test AUC: {ensemble_auc:.4f}")
```

# Neural Network

```python
# Early stopping allows the NN to call back to a version of the model with higher AUC score of an earlier epoch in the NN training

callback = tf.keras.callbacks.EarlyStopping(monitor='val_auc', patience=5, restore_best_weights=True)
```

### Below are several iterations of neural network training, NN4 performed the best

```python
modelNN1 = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='elu'),
    tf.keras.layers.Dense(32, activation='elu'),
    tf.keras.layers.Dense(1, activation = 'sigmoid')
])

modelNN1.compile(
    optimizer = 'adam',
    loss = 'binary_crossentropy',
    metrics=[tf.keras.metrics.AUC(name='auc')]
)
```

```python
history1 = modelNN1.fit(
    x_train_g_scaled, 
    y_train_g,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    callbacks=[callback],
    verbose=1 
)
```
